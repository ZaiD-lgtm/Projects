# initializing a matrix of beta as zero matrix

```python
beta = np.ones((x_new.shape[1],1))
beta.shape
```

## defining sigmoid, gradient descent and cost function

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)  
    return 1 / (1 + np.exp(-z))

def gradient_descent(X, Y, intercept, lr, beta, lambd=0.0):
    m = X.shape[0]
    y_pred = sigmoid(np.dot(X, beta) + intercept)
    
    gradient_beta = (1/m) * np.dot(X.T, (y_pred - Y)) + (lambd / m) * beta
    gradient_intercept = (1/m) * np.sum(y_pred - Y, axis=0)  # Explicit axis

    beta -= lr * gradient_beta
    intercept -= lr * gradient_intercept
    intercept = float(intercept.iloc[0] if isinstance(intercept, np.ndarray) or hasattr(intercept, 'iloc') else intercept)  
    
    return beta, intercept

def cost(X, Y, beta, intercept, lamb=0.0):
    m = X.shape[0]
    y_pred = sigmoid(np.dot(X, beta) + intercept)
    epsilon = 1e-10
    
    cost_val = -(1/m) * np.sum(Y * np.log(y_pred + epsilon) + (1 - Y) * np.log(1 - y_pred + epsilon), axis=0) 
    cost_val += (lamb / (2 * m)) * np.sum(beta ** 2, axis=0)  

    return cost_val
```

```python
print(x_new.shape)
print(y_new.shape)
print(beta.shape)
```

```python
lr = 0.09
iterations = 1000
lamb = 0.5
cost_history = []

for i in range(iterations):
    beta, intercept = gradient_descent(x_new, y_new, intercept, lr, beta,lamb)
    cost_new = cost(x_new, y_new, beta, intercept,lamb)
    cost_history.append(cost_new)
    
    if i % 100 == 0:
        print("Iteration", i, "Cost:", cost_new)

plt.plot(cost_history)
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost Function vs Iterations')
plt.savefig("Costfunction_Graph")
plt.show()

# print("Minimum Cost:", min(cost_history))
```

## Testfile

```python
test = pd.read_csv('Freyja_Pumpkins.csv')
```

## removing unnamed column and mapping both types of pumpkins to 0 and 1

```python
test = test.drop(columns = ['Unnamed: 0'])
test['Class'] = test['Class'].map({'Ürgüp Sivrisi':0 ,'Çerçevelik': 1})
```

## Separation of target and features

```python
test = test.dropna()
test = test.astype(float)
test_features = test.drop(columns = ['Class'])
test_target = test['Class']
```

## Standardization 

```python
means = test_features.mean()
std_div = test_features.std()
std_test = (test_features - means)/std_div
plt.boxplot(std_test,patch_artist = True)
plt.show()
```

```python
# std_test['Class'] = test_target 
test_target = pd.DataFrame(test_target)
test_target = test_target.values.reshape(-1,1)
test_target = pd.DataFrame(test_target)
test_target.shape
```

```python
def predict(X, beta, intercept, threshold=0.5):
    probs = sigmoid(np.dot(X, beta) + intercept)
    predictions = (probs >= threshold).astype(int)
    return predictions
```

```python
y_pred_test = predict(std_test, beta, intercept, threshold=0.5)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(test_target, y_pred_test)
print("Confusion Matrix:")
print(cm)
class_names = ["Ürgüp Sivrisi","Çerçevelik"]
plt.figure(figsize=(10, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="YlGnBu",xticklabels = class_names,yticklabels = class_names)
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix")
plt.show()

print("Accuracy:", accuracy_score(test_target, y_pred_test))
print("Classification Report:\n", classification_report(test_target, y_pred_test))
```
