```python
import random

def add_personas_to_example(example_text, chance=0.60):
    if random.random() < chance:
        persona1 = random.choice(persona1_variants)
        persona2 = random.choice(persona2_variants)
        return f"<bos>persona1: {persona1}\npersona2: {persona2}\nconversation: {example_text}<eos>"
    else:
        return f"<bos>conversation: {example_text}<eos>"
```
